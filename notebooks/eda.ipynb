```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8,6))
```

```python
df = pd.read_csv("../data/cleaned_telco_data.csv")
```

```python
sns.countplot(x='Churn Value', data=df)

plt.title("Churn Distribution")
plt.show()
```

```python
sns.countplot(x='Gender', data=df)

plt.show()
```

```python
sns.countplot(x='Contract', data=df)

plt.xticks(rotation=45)

plt.show()
```

```python
sns.histplot(df['Monthly Charges'],
             bins=30,
             kde=True)

plt.show()
```

```python
sns.histplot(df['Total Charges'],
             bins=30,
             kde=True)

plt.show()
```

```python
sns.histplot(df['Tenure Months'],
             bins=30,
             kde=True)

plt.show()
```

```python
sns.countplot(x='Gender',
              hue='Churn Value',
              data=df)

plt.show()
```

```python
sns.countplot(x='Contract',
              hue='Churn Value',
              data=df)

plt.xticks(rotation=45)

plt.show()
```

```python
sns.countplot(x='Internet Service',
              hue='Churn Value',
              data=df)

plt.show()
```

```python
sns.countplot(x='Partner',
              hue='Churn Value',
              data=df)

plt.show()
```

```python
sns.boxplot(x='Churn Value',
            y='Monthly Charges',
            data=df)

plt.show()
```

```python
sns.boxplot(x='Churn Value',
            y='Total Charges',
            data=df)

plt.show()
```

```python
sns.boxplot(x='Churn Value',
            y='Tenure Months',
            data=df)

plt.show()
```
