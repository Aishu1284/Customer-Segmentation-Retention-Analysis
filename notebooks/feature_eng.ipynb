```python
import pandas as pd
import numpy as np
import pickle

from sklearn.preprocessing import StandardScaler
```

```python
df = pd.read_csv("../data/cleaned_telco_data.csv")
```

```python
df.columns
```

```python
df.drop(columns=['Churn Score','CLTV'],inplace=True)
```

```python
y=df['Churn Value']
```

```python
X=df.drop('Churn Value',axis=1)
```

```python
X = pd.get_dummies(
        X,
        drop_first=True
)
```

```python
X.shape
```

```python
X.columns
```

```python
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

```python
X_scaled = pd.DataFrame(
                X_scaled,
                columns=X.columns
)
```

```python
pickle.dump(
    scaler,
    open("../models/scaler.pkl","wb")
)
```

```python
pickle.dump(
    scaler,
    open("../models/scaler.pkl","wb")
)
```

```python
processed_df = X_scaled.copy()

processed_df['Churn Value'] = y
```

```python
processed_df.to_csv(
    "../data/processed_telco_data.csv",
    index=False
)
```
