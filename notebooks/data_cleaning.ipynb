```python
df['Total Charges'] = df['Total Charges'].replace(' ', np.nan)
```

```python
df['Total Charges'] = pd.to_numeric(df['Total Charges'])
```

```python
df['Total Charges'].fillna(df['Total Charges'].median(),
                          inplace=True)
```

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.drop('CustomerID',
        axis=1,
        inplace=True)
```

```python
df.to_csv("../data/cleaned_telco_data.csv",
          index=False)
```

```python
print(df.columns.tolist())
```

```python
df.drop('Churn Reason', axis=1, inplace=True)
```

```python
cols_to_drop = [
    'Count',
    'Country',
    'State',
    'City',
    'Zip Code',
    'Lat Long',
    'Latitude',
    'Longitude',
    'Churn Label'
]

df.drop(columns=cols_to_drop,
        inplace=True)
```
